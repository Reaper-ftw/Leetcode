```python
from collections import Counter
import heapq


def heapSort(arr,k):
    if k==len(arr):
        return set(arr)
    count=Counter(arr)

    return heapq.nlargest(k,count.keys(),key=count.get)








#Driver code
arr=[1,1,1,1,2,2,2,3]
k=2
result=heapSort(arr,k)
print("the frequent number are",result)
```
