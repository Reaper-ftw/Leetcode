```python
def findMaxandMin(arr,i,j):

  if i==j:
    max_val=arr[i]
    min_val=arr[i]
  elif i==j-1:
    if arr[i]< arr[j]:

      max_val =arr[j]
      min_val=arr[i]
    else:
      max_val =arr[i]
      min_val=arr[j]
  #Big problem divide and conquer approach

  else:
    mid=i+(j-i)//2
    max_l,min_l=findMaxandMin(arr,i,mid)
    max_r,min_r=findMaxandMin(arr,mid+1,j)

    if max_l<max_r:
      max_val=max_r
    else:
      max_val=max_l


    if min_l<min_r:
      min_val=min_l
    else:
      min_val=min_r

  return max_val,min_val
#Driver code

arr=[34,76,10,30,5,80,90,12,-3,-4]

i=0
j=len(arr)-1

max_val,min_val=findMaxandMin(arr,i,j)

print("MAXIMUM AND MINMUM VAKUE OF ARRAY:",max_val,min_val)
```
